```python
import os
import requests
from bs4 import BeautifulSoup
```

```python
url = "https://medium.com/@subashgandyer/papa-what-is-a-neural-network-c5e5cc427c7"

def get_page():
    headers = {
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 Chrome/120.0 Safari/537.36",
        "Accept-Language": "en-US,en;q=0.9"
    }

    try:
        print("Connecting to Medium...")

        response = requests.get(
            url,
            headers=headers,
            timeout=(5, 15)
        )

        response.raise_for_status()

        print("Webpage downloaded successfully.")

        return BeautifulSoup(response.text, "html.parser")

    except requests.exceptions.Timeout:
        print("Error: Medium took too long to respond.")
        return None

    except requests.exceptions.RequestException as e:
        print("Error while accessing Medium:")
        print(e)
        return None
```

```python
def collect_text(soup):

    if soup is None:
        return None

    paragraphs = soup.find_all("p")

    print("Number of paragraphs found:", len(paragraphs))

    text = f"URL: {url}\n\n"

    for paragraph in paragraphs:
        paragraph_text = paragraph.get_text(strip=True)

        if paragraph_text:
            text += paragraph_text + "\n\n"

    return text
```

```python
def collect_text(soup):

    text = f'url: {url}\n\n'

    para_text = soup.find_all('p')

    print(f"paragraphs text = \n {para_text}")

    for para in para_text:
        text += f"{para.text}\n\n"

    return text
```

```python
def save_file(text):

    if text is None:
        print("No text was collected.")
        return

    folder = "scraped_articles"

    if not os.path.exists(folder):
        os.mkdir(folder)

    filename = os.path.join(
        folder,
        "papa-what-is-a-neural-network.txt"
    )

    with open(filename, "w", encoding="utf-8") as file:
        file.write(text)

    print("File saved successfully!")
    print("Location:", os.path.abspath(filename))
```

```python
soup = get_page()

text = collect_text(soup)

save_file(text)
```
